```python
# Ваш код:
```

```python
for i in range(1000):
    if mouse.check_right() == True:
        mouse.turn_right()
        mouse.move()
    else:
        if mouse.check_forward() == True:
            mouse.move()
        else:
            mouse.turn_left()
        
```
